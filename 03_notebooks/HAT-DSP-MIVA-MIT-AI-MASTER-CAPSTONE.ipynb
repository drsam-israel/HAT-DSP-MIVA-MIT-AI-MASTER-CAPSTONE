## Project Aim

To design, implement, and evaluate an AI-enabled healthcare technology assessment and clinical decision support platform that integrates predictive modelling, explainability, model validation, governance, investment assessment, and deployment-readiness evaluation, using 30-day hospital readmission prediction among patients with diabetes as the clinical application.

---

## Clinical AI Use Case

Prediction of recorded 30-day hospital readmission among adults with diabetes using the UCI Diabetes 130-US Hospitals (1999–2008) dataset.

---

## Project Objectives

The specific objectives of this project are to:

1. Develop and retrospectively evaluate machine-learning models for predicting 30-day hospital readmission among patients with diabetes using the UCI Diabetes 130-US Hospitals dataset.

2. Design and implement an interactive readmission risk-scoring interface with model explainability capabilities.

3. Integrate predictive performance, calibration, fairness, and robustness assessments into the platform.

4. Develop structured healthcare technology assessment, AI governance, and scenario-based investment assessment components.

5. Implement an evidence-based deployment-readiness assessment mechanism using predefined technical, clinical, and governance criteria.

6. Evaluate the platform through documented functional, integration, usability, and decision-output testing against predefined requirements and acceptance criteria.

## Project Scope

The project focuses exclusively on 30-day hospital readmission prediction among patients with diabetes using the UCI Diabetes 130-US Hospitals dataset.

The integrated Streamlit platform will provide:

1. Clinical AI risk scoring and model explainability.
2. Predictive performance, calibration, fairness, robustness, and threshold assessment.
3. Clinical utility and healthcare technology assessment.
4. Scenario-based investment decision support.
5. Responsible AI governance and risk assessment.
6. Evidence-based deployment-readiness assessment.
7. Simulated monitoring and audit documentation.

The project does not involve live hospital deployment, operational electronic health record integration, prospective clinical intervention, or regulatory certification.
