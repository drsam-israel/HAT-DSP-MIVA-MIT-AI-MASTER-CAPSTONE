## Project Aim

To design, implement, and evaluate an AI-enabled healthcare technology assessment and clinical decision support platform that integrates predictive modelling, explainability, model validation, governance, investment assessment, and deployment-readiness evaluation, using 30-day hospital readmission prediction among patients with diabetes as the clinical application.

Prediction of recorded 30-day hospital readmission among adults with diabetes using the UCI Diabetes 130-US Hospitals (1999–2008) dataset.

The specific objectives of this project are to:

1. Develop and retrospectively evaluate machine-learning models for predicting 30-day hospital readmission among patients with diabetes using the UCI Diabetes 130-US Hospitals dataset.

3. Integrate predictive performance, calibration, fairness, and robustness assessments into the platform.

4. Develop structured healthcare technology assessment, AI governance, and scenario-based investment assessment components.


### Cell 0.1E — Environment Manifest Preservation

**Purpose:** Record the Python environment, software versions, random seed, and project configuration for reproducibility.

**Method:** Generate a structured JSON environment manifest and save it in the permanent Google Drive project directory.

**Expected Output:** A saved environment manifest with verified content and a SHA-256 integrity checksum.

**Reproducibility:** The manifest documents the computational environment used for HAT-DSP development.

**Version Control:** The master notebook remains under Git version control. Environment configuration will be documented alongside project development.

**Integrity:** The saved manifest will be reloaded and verified without modifying any clinical dataset.


```python

# ============================================================
# HAT-DSP — CELL 0.1E
# ENVIRONMENT MANIFEST PRESERVATION
# ============================================================

# STEP 1: Build the environment manifest.

environment_manifest = {
    "project": "HAT-DSP",
    "programme": "Miva MIT Artificial Intelligence",
    "python_version": sys.version.split()[0],
    "operating_system": platform.system(),
    "platform": platform.platform(),
    "random_seed": RANDOM_SEED,
    "libraries": {
        name: version(package)
        for name, package in packages.items()
    }
}

# STEP 2: Save the manifest to Google Drive.

manifest_path = (
    DRIVE_PROJECT_DIR
    / "00_environment"
    / "environment.json"
)

with open(manifest_path, "w", encoding="utf-8") as file:
    json.dump(environment_manifest, file, indent=4)

# STEP 3: Reload and verify the saved manifest.

with open(manifest_path, "r", encoding="utf-8") as file:
    verified_manifest = json.load(file)

assert verified_manifest == environment_manifest

# STEP 4: Calculate SHA-256 checksum.

manifest_hash = hashlib.sha256(
    manifest_path.read_bytes()
).hexdigest()

# STEP 5: Display verification.

print("=" * 60)
print("HAT-DSP — ENVIRONMENT MANIFEST")
print("=" * 60)

print("Saved file:", manifest_path.name)
print("Python version:", verified_manifest["python_version"])
print("Random seed:", verified_manifest["random_seed"])
print("Libraries recorded:", len(verified_manifest["libraries"]))
print("SHA-256:", manifest_hash)

print("=" * 60)
print("CELL 0.1E — PASSED")
print("ENVIRONMENT MANIFEST SAVED AND VERIFIED")
```
