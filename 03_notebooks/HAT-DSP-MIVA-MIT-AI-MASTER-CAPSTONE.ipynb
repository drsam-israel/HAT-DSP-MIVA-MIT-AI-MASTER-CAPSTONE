
## STAGE 1 — DATASET ACQUISITION AND INTEGRITY VERIFICATION

### Cell 1.1A — Original Dataset ZIP Verification

**Purpose:** Verify the integrity of the original UCI Diabetes 130-US Hospitals dataset archive.

**Method:** Locate the ZIP file in Google Drive, check its file size, calculate its SHA-256 checksum, and test the ZIP archive for corruption.

**Expected Output:** Dataset filename, file size, SHA-256 checksum, archive contents, and verification status.

**Reproducibility:** Record the original dataset checksum for subsequent verification.

**Data Protection:** The original dataset remains in Google Drive and is excluded from the public GitHub repository.

**Integrity:** This cell does not extract, modify, or overwrite the original dataset.


```python

# ============================================================
# HAT-DSP — CELL 1.1A
# ORIGINAL DATASET ZIP INTEGRITY VERIFICATION
# ============================================================

# STEP 1: Locate the original ZIP file.

import zipfile

data_dir = DRIVE_PROJECT_DIR / "01_original_data"

zip_files = list(data_dir.glob("*.zip"))

assert len(zip_files) == 1, (
    f"Expected one ZIP file, found {len(zip_files)}"
)

zip_path = zip_files[0]

# STEP 2: Calculate file size and SHA-256 checksum.

file_size = zip_path.stat().st_size

zip_hash = hashlib.sha256(
    zip_path.read_bytes()
).hexdigest()

# STEP 3: Verify ZIP integrity and inspect contents.

with zipfile.ZipFile(zip_path, "r") as archive:
    corrupted_file = archive.testzip()
    archive_files = archive.namelist()

assert corrupted_file is None, (
    f"Corrupted archive member: {corrupted_file}"
)

# STEP 4: Display verification results.

print("=" * 60)
print("HAT-DSP — ORIGINAL DATASET ZIP VERIFICATION")
print("=" * 60)

print("Filename:", zip_path.name)
print("File size (bytes):", file_size)
print("SHA-256:", zip_hash)

print("\nARCHIVE CONTENTS")

for filename in archive_files:
    print("-", filename)

print("=" * 60)
print("CELL 1.1A — PASSED")
print("ZIP ARCHIVE INTEGRITY VERIFIED")
print("ORIGINAL DATASET NOT MODIFIED")
```


### Cell 1.1B — Dataset Extraction and File Integrity Verification

**Purpose:** Extract the two original CSV files from the verified UCI dataset archive.

**Method:** Extract the files into the original-data directory, calculate their SHA-256 checksums, and verify their integrity against the ZIP archive.

**Expected Output:** Confirmation of both extracted filenames, file sizes, checksums, and integrity status.

**Reproducibility:** Preserve the original ZIP and extracted CSV files without modifying their contents.

**Version Control:** Dataset files remain in Google Drive and are excluded from the public GitHub repository.

**Integrity:** Existing files will not be overwritten unless their contents match the verified archive.


```python

# ============================================================
# HAT-DSP — CELL 1.1B
# DATASET EXTRACTION AND FILE INTEGRITY
# ============================================================

# STEP 1: Define the expected dataset files.

expected_files = [
    "diabetic_data.csv",
    "IDS_mapping.csv"
]

# STEP 2: Verify and extract the original files.

with zipfile.ZipFile(zip_path, "r") as archive:

    assert sorted(archive.namelist()) == sorted(expected_files)

    for filename in expected_files:

        destination = data_dir / filename

        original_bytes = archive.read(filename)

        # Preserve any existing file if it is identical.
        if destination.exists():
            assert destination.read_bytes() == original_bytes, (
                f"Existing file differs: {filename}"
            )
        else:
            destination.write_bytes(original_bytes)

        # Verify extracted file integrity.
        extracted_bytes = destination.read_bytes()

        assert extracted_bytes == original_bytes

        file_hash = hashlib.sha256(
            extracted_bytes
        ).hexdigest()

        print("VERIFIED:", filename)
        print("Size (bytes):", len(extracted_bytes))
        print("SHA-256:", file_hash)
        print("-" * 60)

# STEP 3: Confirm successful extraction.

assert all(
    (data_dir / filename).is_file()
    for filename in expected_files
)

print("=" * 60)
print("CELL 1.1B — PASSED")
print("BOTH DATASET FILES EXTRACTED AND VERIFIED")
print("ORIGINAL ZIP PRESERVED")
print("=" * 60)
```


### Cell 1.3 — Original Dataset Missing-Value Assessment

**Purpose:** Identify missing-value representations and quantify missingness across all 50 original dataset variables.

**Method:** Reload the source CSV as strings while preserving literal values. Count question marks (`?`), the literal text `None`, and empty strings separately.

**Expected Output:** Missing-value totals, variable-level missingness percentages, and a complete data-quality audit.

**Data Integrity:** The original dataset remains unchanged. No missing values are imputed, deleted, or transformed.

**Reproducibility:** Missing-value definitions and calculations are documented explicitly.

**Version Control:** The audit code will be preserved in GitHub. Generated data-quality reports will remain in Google Drive.


```python

# ============================================================
# HAT-DSP — CELL 1.3
# ORIGINAL DATASET MISSING-VALUE ASSESSMENT
# ============================================================

# STEP 1: Reload the original values without automatic NA conversion.

df_literal = pd.read_csv(
    dataset_path,
    dtype="string",
    keep_default_na=False,
    na_filter=False
)

# STEP 2: Count missing-value markers by column.

missing_audit = pd.DataFrame({
    "variable": df_literal.columns,
    "question_mark": (df_literal == "?").sum().values,
    "literal_none": (df_literal == "None").sum().values,
    "empty_string": (df_literal == "").sum().values
})

# STEP 3: Calculate total missingness.

missing_audit["total_missing"] = (
    missing_audit["question_mark"]
    + missing_audit["literal_none"]
    + missing_audit["empty_string"]
)

missing_audit["missing_percent"] = (
    missing_audit["total_missing"]
    / len(df_literal)
    * 100
).round(2)

# STEP 4: Sort variables by missingness.

missing_audit = missing_audit.sort_values(
    "total_missing",
    ascending=False
).reset_index(drop=True)

# STEP 5: Display results.

print("=" * 65)
print("HAT-DSP — ORIGINAL DATASET MISSING-VALUE AUDIT")
print("=" * 65)

print("Total encounters:", len(df_literal))
print("Total variables:", df_literal.shape[1])

print("\nMISSING-VALUE MARKER TOTALS")

print(
    "Question marks:",
    missing_audit["question_mark"].sum()
)

print(
    "Literal None:",
    missing_audit["literal_none"].sum()
)

print(
    "Empty strings:",
    missing_audit["empty_string"].sum()
)

print("\nTOP 10 VARIABLES BY MISSINGNESS")

print(
    missing_audit.head(10).to_string(index=False)
)

# STEP 6: Verify audit completeness.

assert len(missing_audit) == 50

assert (
    missing_audit["total_missing"]
    == missing_audit[
        ["question_mark", "literal_none", "empty_string"]
    ].sum(axis=1)
).all()

print("\n" + "=" * 65)
print("CELL 1.3 — PASSED")
print("MISSING-VALUE AUDIT COMPLETED")
print("NO DATA TRANSFORMATION PERFORMED")
print("=" * 65)
```
