
# Stage 1 — Dataset Acquisition and Integrity Verification

## Stage 1.1 — Dataset Acquisition

### Objective
Acquire the original Diabetes 130-US Hospitals dataset for retrospective 30-day hospital readmission prediction.

### Methodological Approach
Retrieve the original dataset from the UCI Machine Learning Repository and preserve an unmodified copy in the designated project storage directory.

### Relevance to Model Development
Acquisition from an authoritative source establishes data provenance, supports reproducibility, and provides a reliable foundation for subsequent cohort definition, data preparation, and predictive modelling.


```python

# Acquire the original dataset from UCI

from urllib.request import urlretrieve

DATASET_URL = (
    "https://archive.ics.uci.edu/static/public/296/"
    "diabetes+130-us+hospitals+for+years+1999-2008.zip"
)

RAW_DATA_DIR = DRIVE_PROJECT_DIR / "01_original_data"

DATASET_ZIP_PATH = RAW_DATA_DIR / "diabetes_130_us_hospitals.zip"

# Download the original dataset archive
urlretrieve(DATASET_URL, DATASET_ZIP_PATH)

# Verify successful acquisition
assert DATASET_ZIP_PATH.is_file()
assert DATASET_ZIP_PATH.stat().st_size > 0

print("Dataset acquired successfully.")
print("Source: UCI Machine Learning Repository")
print("Dataset ID: 296")
print(f"File name: {DATASET_ZIP_PATH.name}")
print(f"File size: {DATASET_ZIP_PATH.stat().st_size / (1024**2):.2f} MB")
print(f"Storage location: {DATASET_ZIP_PATH}")
```


## Stage 1.2 — Archive Inspection and Extraction

### Objective
Verify the integrity and contents of the downloaded dataset archive before commencing data analysis.

### Methodological Approach
Inspect the ZIP archive, perform an integrity check, and extract the original clinical dataset and associated identifier mapping file into the designated storage directory.

### Relevance to Model Development
Archive verification and preservation of original source files support data provenance, reproducibility, accurate variable interpretation, and reliable downstream analysis.


```python

# Inspect and extract the original dataset archive

import zipfile

with zipfile.ZipFile(DATASET_ZIP_PATH, "r") as archive:

    # Verify archive integrity
    assert archive.testzip() is None, "Archive integrity check failed."

    print("Archive integrity check: PASSED")

    # Inspect archive contents
    print("\nArchive contents:")

    for file in archive.infolist():
        print(
            f"{file.filename}: "
            f"{file.file_size / (1024**2):.2f} MB"
        )

    # Extract original files
    archive.extractall(RAW_DATA_DIR)

# Identify extracted dataset files
DATASET_PATH = RAW_DATA_DIR / "diabetic_data.csv"
MAPPING_PATH = RAW_DATA_DIR / "IDS_mapping.csv"

# Verify required files
assert DATASET_PATH.is_file(), "Clinical dataset not found."
assert MAPPING_PATH.is_file(), "Identifier mapping file not found."

print("\nExtracted files:")
print(f"Clinical dataset: {DATASET_PATH.name}")
print(f"Identifier mappings: {MAPPING_PATH.name}")

print("\nArchive inspection and extraction completed successfully.")
```


## Stage 1.3 — Dataset Loading and Structural Verification

### Objective
Load the original clinical dataset and establish its structural characteristics before undertaking data preparation and analysis.

### Methodological Approach
Import the dataset into a Pandas DataFrame, verify its dimensions, inspect variable names and data types, and confirm the availability of essential identifiers and the readmission outcome variable.

### Relevance to Model Development
Structural verification establishes the initial analytical baseline, identifies potential data inconsistencies, and supports accurate cohort definition and subsequent predictive modelling.


```python

# Load the original clinical dataset
df_raw = pd.read_csv(
    DATASET_PATH,
    keep_default_na=False,
    low_memory=False
)

# Inspect dataset dimensions
print("Dataset loaded successfully.")
print(f"Total encounters: {df_raw.shape[0]:,}")
print(f"Total variables: {df_raw.shape[1]}")

# Verify expected dataset dimensions
assert df_raw.shape == (101766, 50), (
    "Unexpected dataset dimensions."
)

# Verify essential variables
required_columns = [
    "encounter_id",
    "patient_nbr",
    "readmitted"
]

assert all(
    column in df_raw.columns
    for column in required_columns
), "Essential variables are missing."

# Verify identifier mapping file
assert MAPPING_PATH.stat().st_size > 0

print(f"Mapping file size: {MAPPING_PATH.stat().st_size:,} bytes")

# Inspect dataset structure
print("\nDataset structure:")
df_raw.info()

# Summarise variable data types
print("\nData type distribution:")
display(
    df_raw.dtypes.astype(str)
    .value_counts()
    .rename_axis("Data Type")
    .reset_index(name="Variable Count")
)

print("\nDataset structural verification: PASSED")
```


## Stage 1.4 — Dataset Integrity and Provenance Verification

### Objective
Establish the integrity, traceability, and identifier consistency of the original clinical dataset.

### Methodological Approach
Examine encounter uniqueness, patient identifiers, duplicate records, and valid readmission outcome categories. Generate SHA-256 checksums and preserve a dataset provenance record.

### Relevance to Model Development
Dataset integrity verification supports reliable cohort construction, patient-aware data partitioning, reproducibility, and prevention of identifier-related analytical errors.


```python

# Examine dataset integrity

total_encounters = len(df_raw)
unique_encounters = df_raw["encounter_id"].nunique()
unique_patients = df_raw["patient_nbr"].nunique()

duplicate_rows = int(df_raw.duplicated().sum())

duplicate_encounters = int(
    df_raw["encounter_id"].duplicated().sum()
)

missing_identifiers = int(
    df_raw[["encounter_id", "patient_nbr"]]
    .isna().sum().sum()
)

# Verify readmission outcome categories
expected_outcomes = {"<30", ">30", "NO"}
observed_outcomes = set(df_raw["readmitted"].unique())

# Generate source file checksums
source_files = [
    DATASET_ZIP_PATH,
    DATASET_PATH,
    MAPPING_PATH
]

file_checksums = {
    file.name: hashlib.sha256(file.read_bytes()).hexdigest()
    for file in source_files
}

# Create dataset provenance record
dataset_provenance = {
    "dataset_name": "Diabetes 130-US Hospitals (1999-2008)",
    "source": "UCI Machine Learning Repository",
    "dataset_id": 296,
    "source_url": DATASET_URL,
    "total_encounters": total_encounters,
    "total_variables": df_raw.shape[1],
    "unique_encounters": unique_encounters,
    "unique_patients": unique_patients,
    "duplicate_rows": duplicate_rows,
    "duplicate_encounter_ids": duplicate_encounters,
    "missing_identifiers": missing_identifiers,
    "readmission_categories": sorted(observed_outcomes),
    "sha256_checksums": file_checksums
}

# Save provenance record
provenance_path = RAW_DATA_DIR / "dataset_provenance.json"

with open(provenance_path, "w") as file:
    json.dump(dataset_provenance, file, indent=4)

# Display integrity results
print("DATASET INTEGRITY SUMMARY")
print("-" * 45)

print(f"Total encounters: {total_encounters:,}")
print(f"Unique encounter IDs: {unique_encounters:,}")
print(f"Unique patients: {unique_patients:,}")
print(f"Duplicate rows: {duplicate_rows:,}")
print(f"Duplicate encounter IDs: {duplicate_encounters:,}")
print(f"Missing identifiers: {missing_identifiers:,}")
print(f"Readmission categories: {sorted(observed_outcomes)}")

# Verify integrity requirements
assert duplicate_rows == 0, "Duplicate records detected."
assert duplicate_encounters == 0, "Duplicate encounter IDs detected."
assert missing_identifiers == 0, "Missing identifiers detected."
assert observed_outcomes == expected_outcomes, "Unexpected outcome categories."

print("\nSHA-256 CHECKSUMS")

for filename, checksum in file_checksums.items():
    print(f"{filename}: {checksum}")

print(f"\nProvenance record saved: {provenance_path}")
print("\nDataset integrity verification: PASSED")
```


## Stage 1.5 — Dataset Understanding and Variable Classification

### Objective
Establish a structured understanding of the original dataset by classifying variables according to their clinical, administrative, and analytical roles.

### Methodological Approach
Group variables into meaningful categories, document their data types and analytical roles, and generate a preliminary data dictionary for subsequent analysis.

### Relevance to Model Development
Systematic variable classification supports clinical interpretation, cohort definition, appropriate preprocessing, predictor selection, and prevention of data leakage.


```python

# Classify dataset variables by functional category

variable_groups = {
    "Identifiers": [
        "encounter_id", "patient_nbr"
    ],

    "Demographics": [
        "race", "gender", "age", "weight"
    ],

    "Admission and Discharge": [
        "admission_type_id",
        "discharge_disposition_id",
        "admission_source_id",
        "time_in_hospital"
    ],

    "Administrative Information": [
        "payer_code", "medical_specialty"
    ],

    "Clinical Procedures and Medication Counts": [
        "num_lab_procedures",
        "num_procedures",
        "num_medications"
    ],

    "Previous Healthcare Utilisation": [
        "number_outpatient",
        "number_emergency",
        "number_inpatient"
    ],

    "Clinical Diagnoses": [
        "diag_1", "diag_2", "diag_3",
        "number_diagnoses"
    ],

    "Laboratory Results": [
        "max_glu_serum", "A1Cresult"
    ],

    "Diabetes Medications": [
        "metformin", "repaglinide", "nateglinide",
        "chlorpropamide", "glimepiride",
        "acetohexamide", "glipizide", "glyburide",
        "tolbutamide", "pioglitazone",
        "rosiglitazone", "acarbose", "miglitol",
        "troglitazone", "tolazamide", "examide",
        "citoglipton", "insulin",
        "glyburide-metformin", "glipizide-metformin",
        "glimepiride-pioglitazone",
        "metformin-rosiglitazone",
        "metformin-pioglitazone"
    ],

    "Treatment Management": [
        "change", "diabetesMed"
    ],

    "Outcome": [
        "readmitted"
    ]
}

# Create preliminary data dictionary
dictionary_records = []

for category, variables in variable_groups.items():
    for variable in variables:

        if category == "Identifiers":
            role = "Identifier"
        elif category == "Outcome":
            role = "Source outcome variable"
        else:
            role = "Candidate predictor or assessment variable"

        dictionary_records.append({
            "Variable": variable,
            "Category": category,
            "Data Type": str(df_raw[variable].dtype),
            "Analytical Role": role,
            "Unique Values": df_raw[variable].nunique()
        })

data_dictionary = pd.DataFrame(dictionary_records)

# Verify complete variable classification
assert len(data_dictionary) == df_raw.shape[1]
assert data_dictionary["Variable"].is_unique
assert set(data_dictionary["Variable"]) == set(df_raw.columns)

# Summarise variable categories
variable_summary = (
    data_dictionary.groupby("Category", sort=False)
    .size()
    .reset_index(name="Variable Count")
)

print("DATASET VARIABLE CLASSIFICATION")
display(variable_summary)

print(f"Total variables classified: {len(data_dictionary)}")
print(f"Variable categories: {len(variable_groups)}")

# Display preliminary data dictionary
print("\nPRELIMINARY DATA DICTIONARY")
display(data_dictionary)

# Save data dictionary and classification summary
data_dictionary.to_csv(
    RAW_DATA_DIR / "data_dictionary.csv",
    index=False
)

variable_summary.to_csv(
    RAW_DATA_DIR / "variable_classification_summary.csv",
    index=False
)

print("\nData dictionary and classification summary saved.")
print("Variable classification: PASSED")
```
