
## Stage 0.5.1 — Local Git Repository Initialisation

### Objective
Establish version control for tracking project source code, configuration files, notebooks, and technical documentation.

### Methodological Approach
Initialise a Git repository with a main branch and configure exclusion rules to prevent sensitive data, credentials, and large generated artifacts from entering version control.

### Relevance to System Development
Version control supports development traceability, reproducibility, controlled changes, and professional software lifecycle management.


```python

# Import Git execution utility
import subprocess

# Define local Git repository
GIT_REPO_DIR = Path(
    "/content/HAT-DSP-MIVA-MIT-AI-MASTER-CAPSTONE"
)

GIT_REPO_DIR.mkdir(parents=True, exist_ok=True)

# Initialise Git repository
subprocess.run(
    ["git", "init", "-b", "main"],
    cwd=GIT_REPO_DIR,
    check=True
)

# Define Git exclusion rules
gitignore_content = """
# Datasets
01_original_data/
02_processed_data/
*.csv
*.parquet

# Generated models and artifacts
04_models/
*.pkl
*.joblib

# Logs and temporary files
11_logs/
*.log
__pycache__/
.ipynb_checkpoints/
.pytest_cache/

# Credentials and secrets
.env
.env.*
!.env.example
.streamlit/secrets.toml
*.pem
*.key

# Virtual environments
venv/
.venv/
"""

# Save Git exclusion rules
(GIT_REPO_DIR / ".gitignore").write_text(
    gitignore_content.strip() + "\n"
)

# Verify Git configuration
print("Local Git repository:", GIT_REPO_DIR)

subprocess.run(["git", "--version"], check=True)

subprocess.run(
    ["git", "status", "--short", "--branch"],
    cwd=GIT_REPO_DIR,
    check=True
)

print("\nGit exclusion rules configured successfully.")
print("Local repository initialisation completed.")
```

### Objective
Establish a connection between the local project repository and its remote GitHub repository.

### Methodological Approach
Configure the remote repository URL and verify the Git connection settings before synchronising the initial project commit.

### Relevance to System Development
Remote version control supports persistent development history, academic transparency, reproducibility, and independent review of project source code.


```python

# Configure remote GitHub repository

GITHUB_USERNAME = input("drsam-israel: ").strip()

REPO_NAME = "HAT-DSP-MIVA-MIT-AI-MASTER-CAPSTONE"

REMOTE_URL = (
    f"https://github.com/drsam-israel/HAT-DSP-MIVA-MIT-AI-MASTER-CAPSTONE.git"
)

# Check existing remote configuration
remotes = subprocess.run(
    ["git", "remote"],
    cwd=GIT_REPO_DIR,
    capture_output=True,
    text=True,
    check=True
).stdout.split()

# Configure remote repository
if "origin" in remotes:
    command = ["git", "remote", "set-url", "origin", REMOTE_URL]
else:
    command = ["git", "remote", "add", "origin", REMOTE_URL]

subprocess.run(
    command,
    cwd=GIT_REPO_DIR,
    check=True
)

# Verify remote configuration
result = subprocess.run(
    ["git", "remote", "-v"],
    cwd=GIT_REPO_DIR,
    capture_output=True,
    text=True,
    check=True
)

print(result.stdout)
print("Remote GitHub repository configured successfully.")
```
